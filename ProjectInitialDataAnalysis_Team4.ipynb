### Data Visualizations

</div>

An exploratory data visualization was done for better understanding of the data, look at outliers, skewness and get a sense of relationship within the predictors or of the predictors with the target variable. Several visualizations were looked at and those that are more useful to understand the problem are presented here.

<div style="text-align: center;">
    <img src="images/loanIntent.png" alt="loantype" />
</div>

This pie chart shows us the distribution of the purpose of the loan that was applied for. The highest percentage of loan applications were related to education, followed by medical reasons, centure, personal reasons, debt consolidation and lowest category was for home improvement. This helps us decide how many levels to keep for this variable when modeling.

<div style="text-align: center;">
    <img src="images/outliers.png" alt="outlier" />
</div>

This graph shows the relationship between applicant's employment period and age in years. Overall the distribution looks normal with some obvious extremities. One extremity is where the applicant's age is significantly higher and with low employment period. Right away, this doesn't call for any actions as these scenarios are possible in the real world, but it is the other extremity that one should be worried about. The second extremity is where a person's age is below 25 years, but their employment length is above 120 years. Now, this is an impossible situation to occur in reality and therefore, it makes sense to remove any and all instances where an applicant's age is smaller than their employment length.  
More about removing these observations is discussed in the outliers section.

<div style="text-align: center;">
    <img src="images/percentIncome_homeOwnership.png" alt="homeOwners" />
</div>

The above graph explores the relationship between the percentage of income allowed towards loan repayment and home ownership type of the applicant.
These are two important variables that impact the target variable and the same is seen in the graph above. There is a clear relationship between the percentage of income allocated and loan approval status. In general, higher the allocated amount for each home ownership type, higher the chances for loan approval. There is also a direct relationship between the percent of income allocated versus home ownership type. In general, applicant's that either own or rent the home or those who fall under the `OTHER` category which could be living out a motel/van/airbnb, etc. are more likely to allocate more money towards loan replayment and hence have higher chances of getting them approved as well.

<div style="text-align: center;">
    <img src="images/LoanApproval_Intent.png" alt="loanApprovalVsIntent" />
</div>

This bar distribution shows the relationship between loan approval rate versus the intent of the loan. In general, there is a higher approval rate for loan corresponding to debt consolidation. Debt consolidation combines multiple debts into a single loan type and makes it easier to pay off debt, so having a higher approval rate for that makes sense. The second highest approval rate is for loans corresponding to medical reasons followed closely by home improvement. The next in line are personal loans, education loans and then the loan corresponding to ventures. Again, it makes sense that these three have lower approval rate in comparison to others as they fall under a certain risk category, particularly the venture loans where are for high-growth companies that are backed by venture capital, but if there is always a certain risk associated with the success of business in the early-stage.

<div style="text-align: center;">
    <img src="images/loanApprovalRelations.png" alt="loanApprovalVsVariables" />
</div>

The above set of plots is useful to understand the relationship between the target variables and predictors. This gives an idea of what kind of relationship each predictor has with the target variables and how strong or weak that relationship is.  
For applicants that either rent the home or fall under `Other` category, the loan approval rate is much higher compared to other two categories. Those who own home have lower approval rates, which is counterintiuitive but these kind of complex relationships are what are hard to predict without doing the thorough analysis.  
Loan applications for debt consolidation, medical reasons and home improvements have higher approval rates compared to the personal loans, education loans and venture loans.  
Lower the loan grade lower is the risk, however there is a negative correlation between the risk and loan approval. Higher the grade, higher the risk and also higher interest rate(from heatmap for correlation) and higher the chances for loan approval. in fact for grade G, the rate of rejection is very low.  
If a person has a default history, their chances of loan approval are higher which is again counterintuitive. Hence, this visualization is useful to study that default history has direct impact on loan approval. 

<div style="text-align: center;">
    <img src="images/corrMatrix.png" alt="correlation" />
</div>

This is a heatmap looking at correlations between all the variables if the abosolute value of their correlation is greater than 0.5. This gives a quick overview of variables that are highly correlated with each other.  
There is a high positive correlation between credit history length and age of the applicant, loan amount and amount of income allocated towards loan replayment, applicant's default history and loan interest rate.
There is a high negative correlation between loan grade A (i.e. low risk) and loan interest rate, no default history and loan interest rate, home ownership type of mortage and rent.

### Handling Missing Values

</div>

Only two variables `person_emp_length` (2.75%) and `loan_int_rate` (9.56%) have missing values. The missing values were imputed using the `kNN` technique by looking at 5 nearest neighbors using the Euclidian distance from the remaining non-missing values for those variables.

<div style="text-align: center; margin-top: 50px;">

### Outliers

</div>

Doing a quick data visualization, it was realized that there are some cases where a person's employment length in years is much much bigger than their actual age. This obviously doesn't make sense. A logical vector was generated to check for such instances and two such instances were found. Since, it doesn't make sense to keep these events, the rows containing these outliers were removed from the dataset entirely. 

Graph showing these two outliers is presented in the visualizations section above.

### Dummy Variables/ Factors

</div>

In total, there are four categorical variables converted to factors with various levels. These levels increases the dimensions for input variables used for modelling. Hence, the number of levels were reduced using `fct_lump` to a reasonable number of levels that would make the most impact on the target variables. Levels were reduced for the variables `loan_grade` and `loan_intent` only. 

<div style="text-align: center; margin-top: 50px;">

### Feature transformation

</div>

Transoformed the variable `person_income` using logarithmic transformation. The distribution for this variable was right skewed and there are some instances where the income is very high compared to the average income of the applicants, hence it made sense to make this transformation.
